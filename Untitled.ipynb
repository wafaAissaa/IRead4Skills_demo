```python
def replace_NA(obj):
    if isinstance(obj, dict):
        return {k: replace_NA(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [replace_NA(elem) for elem in obj]
    elif isinstance(obj, str) and obj == "NA":
        return "na"
    else:
        return obj
```

```python
def find_full_key_path(d, target_key, path=None):
    # gets the path of the feature in the outputs.json
    if path is None:
        path = []
    if isinstance(d, dict):
        for k, v in d.items():
            new_path = path + [k]
            if k == target_key:
                return new_path
            result = find_full_key_path(v, target_key, new_path)
            if result is not None:
                return result
    return None
```

```python
import os
import json

folder = "./outputs"  # change this to your folder path
files_with_na = []

def contains_NA(obj):
    """Recursively check if obj contains the exact value 'NA' (string)."""
    if isinstance(obj, dict):
        return any(contains_NA(v) for v in obj.values())
    elif isinstance(obj, list):
        return any(contains_NA(v) for v in obj)
    else:
        return obj == "NA"  # exact match only

for filename in os.listdir(folder):
    if filename.endswith(".json"):
        filepath = os.path.join(folder, filename)
        try:
            with open(filepath, "r", encoding="utf-8") as f:
                data = json.load(f)
            if contains_NA(data):
                files_with_na.append(filename)
        except Exception as e:
            print(f"⚠️ Could not read {filename}: {e}")

print("Files containing exact 'NA':")
for f in files_with_na:
    print(f)
```
